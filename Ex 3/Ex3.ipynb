Driver Standings Dataset

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load Dataset
file_path = "driver_standings(1).csv"
df = pd.read_csv(
    file_path,
    low_memory=False
)

# Select Important Columns
standings_df = pd.DataFrame({
    "DriverStandingsId": df["driverStandingsId"],
    "RaceId": df["raceId"],
    "DriverId": df["driverId"],
    "Points": df["points"],
    "Position": df["position"],
    "PositionText": df["positionText"],
    "Wins": df["wins"]
})

print("Original DataFrame:")
display(standings_df.head(5))
````

````python
# 1. First 5 Rows
print("\n1. First 5 rows:")
display(standings_df.head())

# 2. Last 2 Rows
print("\n2. Last 2 rows:")
display(standings_df.tail(2))

# 3. Shape
print("\n3. Shape:")
print(standings_df.shape)
````

````python
# 4. Column Names
print("\n4. Column Names:")
print(standings_df.columns)

# 5. Data Types
print("\n5. Data Types:")
print(standings_df.dtypes)

# 6. Statistical Summary
print("\n6. Statistical Summary:")
display(standings_df.describe())
````

````python
# 7. Missing Values
print("\n7. Missing Values:")
display(standings_df.isnull().sum())

# 8. Driver standings with more than 100 points
print("\n8. Driver standings with more than 100 points:")
high_points = standings_df[
    standings_df["Points"] > 100
][["DriverId", "Points", "Position", "Wins"]].head(10)
display(high_points)
````

````python
# 9. First-position standings
print("\n9. First-position standings:")
first_position = standings_df[
    standings_df["Position"] == 1
][["DriverId", "Points", "Position", "Wins"]].head(10)
display(first_position)
````

````python
# 10. Visualization - Points
points_all = standings_df.dropna(subset=["Points"])

plt.figure(figsize=(10, 5))
plt.hist(points_all["Points"], bins=20)
plt.title("Points Distribution")
plt.xlabel("Points")
plt.ylabel("Number of Standings Records")
plt.tight_layout()
plt.show()
````

````python
# 11. Position Distribution
position_counts = standings_df["Position"].value_counts().sort_index().head(15)

plt.figure(figsize=(10, 5))
plt.bar(position_counts.index.astype(str), position_counts.values)
plt.title("Top Finishing Positions")
plt.xlabel("Finishing Position")
plt.ylabel("Number of Standings Records")
plt.tight_layout()
plt.show()
````

````python
# 12. Wins Distribution
wins_counts = standings_df["Wins"].value_counts().sort_index().head(15)

plt.figure(figsize=(10, 5))
plt.bar(wins_counts.index.astype(str), wins_counts.values)
plt.title("Wins Distribution")
plt.xlabel("Wins")
plt.ylabel("Number of Standings Records")
plt.tight_layout()
plt.show()
````

````python
# 13. Correlation Matrix
print("\n13. Correlation Matrix:")
display(standings_df.corr(numeric_only=True))
````
