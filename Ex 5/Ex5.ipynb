# EXPT NO:5A — DATA VISUALIZATION USING MATPLOTLIB
## Driver Standings Dataset

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("driver_standings(1).csv", low_memory=False)
driver_points = df.groupby("driverId")["points"].max().sort_values(ascending=False).head(10)

# Program 1: Line Chart
plt.plot(driver_points.index.astype(str), driver_points.values, marker="o")
plt.title("Maximum Points by Top Driver IDs")
plt.xlabel("Driver ID")
plt.ylabel("Maximum Points")
plt.grid()
plt.show()

# Program 2: Bar Chart
plt.bar(driver_points.index.astype(str), driver_points.values)
plt.title("Top Driver IDs by Maximum Points")
plt.xlabel("Driver ID")
plt.ylabel("Maximum Points")
plt.xticks(rotation=45)
plt.show()
````

````python
# Program 3: Pie Chart
plt.pie(driver_points.values, labels=driver_points.index.astype(str), autopct="%1.1f%%")
plt.title("Top 10 Driver IDs - Points Share")
plt.show()

# Program 4: Histogram
plt.hist(df["points"].dropna(), bins=15)
plt.title("Driver Points Distribution")
plt.xlabel("Points")
plt.ylabel("Frequency")
plt.show()

# Program 5: Scatter Plot
plt.scatter(df["points"], df["wins"], s=12)
plt.title("Points vs Wins")
plt.xlabel("Points")
plt.ylabel("Wins")
plt.show()
````

````python
# Program 6: Box Plot
plt.boxplot(df["points"].dropna(), vert=False)
plt.title("Driver Points Box Plot")
plt.xlabel("Points")
plt.show()

# Program 7: Multiple Line Charts
top5 = driver_points.head(5)
top5_wins = df.groupby("driverId")["wins"].max().reindex(top5.index)
x = range(len(top5))
plt.plot(x, top5.values, marker="o", label="Maximum Points")
plt.plot(x, top5_wins.values, marker="o", label="Maximum Wins")
plt.title("Maximum Points and Wins - Top Driver IDs")
plt.xlabel("Driver ID")
plt.ylabel("Value")
plt.xticks(list(x), top5.index.astype(str))
plt.legend()
plt.grid()
plt.show()
````

# EXPT NO:5B — DATA VISUALIZATION
## Driver Standings Dataset

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("driver_standings(1).csv", low_memory=False)

print("Dataset Preview:")
display(df.head())

num_cols = df.select_dtypes(include="number").columns
print("\nNumerical columns:")
print(num_cols.tolist())

# 1. BAR CHART
top10 = df.groupby("driverId")["wins"].max().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
plt.bar(top10.index.astype(str), top10.values)
plt.title("Driver Standings - Top 10 Driver IDs by Maximum Wins")
plt.xlabel("Driver ID")
plt.ylabel("Maximum Wins")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

````python
# 2. LINE CHART
first20 = df.head(20)
plt.figure(figsize=(9,5))
plt.plot(range(1, len(first20)+1), first20["points"], marker="o")
plt.title("Driver Standings - Points of First 20 Records")
plt.xlabel("Record")
plt.ylabel("Points")
plt.grid()
plt.tight_layout()
plt.show()

# 3. PIE CHART
plt.figure(figsize=(7,7))
plt.pie(top10, labels=top10.index.astype(str), autopct="%1.1f%%")
plt.title("Driver Standings - Top 10 Wins Share")
plt.tight_layout()
plt.show()
````

````python
# 4. BOX PLOT
plt.figure(figsize=(8,5))
sns.boxplot(y=df["wins"])
plt.title("Driver Standings - Wins Box Plot")
plt.ylabel("Wins")
plt.tight_layout()
plt.show()

# 5. HISTOGRAM
plt.figure(figsize=(8,5))
plt.hist(df["wins"], bins=15)
plt.title("Driver Standings - Wins Histogram")
plt.xlabel("Wins")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
````
