# DVT 4 — EDA: Data Inspection, Filtering, Statistical Analysis and Correlation

Driver Standings Dataset

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("driver_standings(1).csv")

# Display dataset
print(df)

# Display first five records
print(df.head())

# Display shape
print(df.shape)

# Display data types
print(df.dtypes)
````

````python
# Display statistical summary
print(df.describe())

# Filter standings with points greater than 100
print(df[df["points"] > 100])

# Filter standings with finishing position 1
print(df[df["position"] == 1])

# Sort standings by points
print(df.sort_values("points", ascending=False))

# Sort standings by wins
print(df.sort_values("wins", ascending=False))

# Driver-wise maximum points
print(df.groupby("driverId")["points"].max())
````

````python
# Driver-wise maximum wins
print(df.groupby("driverId")["wins"].max())

# Calculate mean
print("Mean Points:", df["points"].mean())

# Calculate median
print("Median Points:", df["points"].median())

# Calculate minimum
print("Minimum Points:", df["points"].min())

# Calculate maximum
print("Maximum Points:", df["points"].max())

# Calculate standard deviation
print("Standard Deviation:", df["points"].std())

# Display correlation matrix
print(df[["points", "wins"]].corr())
````

````python
# Display correlation between points and wins
print("Correlation:", df["points"].corr(df["wins"]))

# Bar chart
top10 = df.groupby("driverId")["points"].max().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))
plt.bar(top10.index.astype(str), top10.values)
plt.title("Top 10 Driver IDs by Maximum Points")
plt.xlabel("Driver ID")
plt.ylabel("Maximum Points")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Line chart
first20 = df.head(20)
plt.figure(figsize=(10, 5))
plt.plot(range(1, len(first20) + 1), first20["points"], marker="o")
plt.title("Points of First 20 Driver Standings Records")
plt.xlabel("Record")
plt.ylabel("Points")
plt.show()
````

````python
# Scatter plot
plt.figure(figsize=(10, 5))
plt.scatter(df["points"], df["wins"])
plt.title("Points vs Wins")
plt.xlabel("Points")
plt.ylabel("Wins")
plt.tight_layout()
plt.show()

# Histogram
plt.figure(figsize=(10, 5))
plt.hist(df["points"].dropna(), bins=20)
plt.title("Points Distribution")
plt.xlabel("Points")
plt.ylabel("Number of Standings Records")
plt.tight_layout()
plt.show()

# Box plot
plt.figure(figsize=(8, 5))
plt.boxplot(df["points"].dropna(), vert=False)
plt.title("Points Box Plot")
plt.xlabel("Points")
plt.tight_layout()
plt.show()
````

````python
# Finishing position distribution
position_counts = df["position"].value_counts().sort_index().head(15)

plt.figure(figsize=(10, 5))
plt.bar(position_counts.index.astype(str), position_counts.values)
plt.title("Finishing Position Distribution")
plt.xlabel("Position")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()

# Missing values
print("Missing Values:")
print(df.isnull().sum())

# Duplicate rows
print("Duplicate Rows:", df.duplicated().sum())

# Numerical correlation matrix
print(df.corr(numeric_only=True))
````

# 4B — Exploratory Data Analysis and Visualization

````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load Driver Standings dataset
df = pd.read_csv("driver_standings(1).csv")

# Display dataset
print("Dataset Preview:")
display(df.head())

# Select numerical columns
num_cols = df.select_dtypes(include="number").columns

print("\nNumerical columns:")
print(num_cols.tolist())
````

````python
# BAR CHART
top10 = df.groupby("driverId")["points"].max().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))
plt.bar(top10.index.astype(str), top10.values)
plt.title("Top 10 Driver IDs by Maximum Points")
plt.xlabel("Driver ID")
plt.ylabel("Maximum Points")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# BOX PLOT
plt.figure(figsize=(8, 5))
plt.boxplot(df["points"].dropna(), vert=False)
plt.title("Box Plot of Driver Points")
plt.xlabel("Points")
plt.tight_layout()
plt.show()
````

````python
# LINE CHART
first20 = df.head(20)

plt.figure(figsize=(10, 5))
plt.plot(range(1, len(first20) + 1), first20["points"], marker="o")
plt.title("Points of First 20 Driver Standings Records")
plt.xlabel("Record")
plt.ylabel("Points")
plt.grid()
plt.tight_layout()
plt.show()
````
