# DVT 2 – EDA – Data Cleaning

Driver Standings Dataset

````python
import os
import pandas as pd
import sqlite3

# Dataset path
path = "/content"
print("Dataset path:", path)
print("\nFiles in dataset:")
print(os.listdir(path))

# Find CSV file automatically
csv_files = [f for f in os.listdir(path) if f.lower().endswith(".csv")]
if not csv_files:
    raise FileNotFoundError("No CSV file found.")

csv_file = "driver_standings(1).csv"
print("\nCSV file:", csv_file)

# Load dataset
df = pd.read_csv(
    os.path.join(path, csv_file),
    low_memory=False
)

print("\nFirst 5 rows:")
display(df.head())
````

````python
print("\nLast 5 rows:")
display(df.tail())

print("\nDataset information:")
df.info()

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
````

````python
print("\nStatistical description:")
display(df.describe(include="all").T)

print("\nMissing values:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate rows:")
print(df.duplicated().sum())
````

````python
print("\nTop 10 Driver IDs:")
display(df["driverId"].value_counts().head(10))

print("\nFinishing Positions:")
display(df["position"].value_counts().sort_index().head(10))

print("\nCorrelation:")
display(df.corr(numeric_only=True))
````

````python
# Export CSV
df.to_csv("driver_standings.csv", index=False)

# Export Excel
df.to_excel("driver_standings.xlsx", index=False)

# SQLite database
conn = sqlite3.connect("driver_standings.db")
df.to_sql(
    "driver_standings",
    conn,
    if_exists="replace",
    index=False
)
conn.close()

print("\nCSV, Excel and SQLite database files created successfully.")
````
